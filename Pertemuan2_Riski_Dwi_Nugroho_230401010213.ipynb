```python
import pandas as pd
import numpy as np

URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
```

```python
df = pd.read_csv(URL)
df.head()
```

```python
# 1. Jumlah Baris 891 dan jumlah kolom 12
# 2. Pada kolom Cabin beberapa data yang missing seperti adanya value NaN
print(df.shape, df.columns.tolist())
```

```python
# 3. Total Persentase penumpang yang selamat adalah 74% demgan rata rata umur 27 tahun

print(df['Survived'].mean(), df['Age'].mean())
```

```python
# 4. Persentase Wanita yang selamat pada Kelas 1 adalah 96% dengan rata rata umur 34 tahun

wk1 = df[(df['Sex']=='female')&
         (df['Pclass']==1)]
print(wk1['Survived'].mean(), wk1['Age'].mean())
```

```python
# 5. Kelas dengan tingkat keselamatan tertinggi dengan dihitung dari persentse adalah Kelas 1

hasil = df.groupby('Pclass').agg(
    total=('PassengerId', 'count'),
    selamat=('Survived','mean')
).round(3)
print(hasil)
```

```python
# 6. Tampilkan Hasil

hasil2 = df.groupby('Pclass')['Survived'].mean()
print(hasil2)
```
