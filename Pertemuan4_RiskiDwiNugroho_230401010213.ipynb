```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
from scipy import stats
```

```python
# 1. Load & Inspect Dataset
df = sns.load_dataset('iris')
print('Shape:', df.shape)
print(df.dtypes)
print(df.head())
print(df.describe().round(3))
```

```python
# 2. Menghtitung Mean, median, std, varians skewness, kurtosis
for col_name in df.select_dtypes(include='number').columns:
  col = df[col_name]
  print(f'\n=== {col_name} ===')
  print(f' Mean : {col.mean() : .3f}')
  print(f' Median : {col.median():.3f}')
  print(f' Std  : {col.std():.3f}')
  print(f' Skewness : {col.skew():.3f}')
  print(f' Kurtosis : {col.kurtosis():.3f}')
```

```python
# 3. Analisis Distribusi
fig, ax = plt.subplots(figsize=(8, 4))
sns.histplot(df['sepal_length'], kde=True,
            color='#028090', bins=20, ax=ax)
ax.axvline(df['sepal_length'].mean(), color='red', linestyle='dashed', linewidth=2, label=f"Mean{df['sepal_length'].mean():.2f}")
ax.axvline(df['sepal_length'].median(), color='orange', linestyle='dashed', linewidth=2, label=f"Median{df['sepal_length'].median():.2f}")

ax.set_title('Distribution Sepal Length');
ax.legend();
plt.show()
```

```python
# 4. Boxplot per Spesies
flg, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.boxplot(x='species', y='petal_length', data=df, ax=axes[0], palette='Set2')
axes[0].set_title('Boxplot Petal Length')

sns.violinplot(x='species', y='petal_length', data=df, ax=axes[1], palette='Set2')
axes[1].set_title('Violinplot Petal Length')

plt.tight_layout()
plt.show()
```

```python
# 5. Matriks Korelasi Pearson
corr = df.drop('species', axis=1).corr(method='pearson')
print(corr.round(3))

import numpy as np
mask = np.triu(np.ones(corr.shape, dtype=bool))
corr_masked = corr.where(~mask)
max_pair = corr_masked.stack().idxmax()
print(f'Korelasi tertinggi: {max_pair}={corr.loc[max_pair]:.3f}')
```

```python
# 6. Scatter Plot & Heatmap
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species', palette='Set2', ax=axes[0])
sns.regplot(data=df, x='petal_length', y='petal_width', scatter=False, color='gray', ax=axes[0])
axes[0].set_title('Scatter: Petal Length vs Petal Width')
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu_r', vmin=-1, vmax=1, square=True, ax=axes[1])
axes[1].set_title('Heatmap Korelasi')
plt.tight_layout();
plt.show()
```
